```python
import numpy as np
```

```python
X=np.array([[0,0],[0,1],[1,0],[1,1]])
y=np.array([0,1,1,0])
X,y
```

```python
def step_function(z):
    return 1 if z>=0 else 0
```

```python
x1=0
x2=0

w1=1
w2=1
w3=1
w4=1

b1=-0.5
b2=-1.5
b3=-0.5
```

```python
def forward_propogation(x1,x2):
    w1=1
    w2=1
    w3=1
    w4=1

    b1=-0.5
    b2=-1.5
    b3=-0.5
    z1=w1*x1+w3*x2+b1
    z2=w2*x1+w4*x2+b2

    h1=step_function(z1)
    h2=step_function(z2)
    w5=1
    w6=-2
    z3=w5*h1+w6*h2+b3
    h3=step_function(z3)
    return h3
```

```python
print("XOR results using ANN forward propogation:")
for x in X:
    print(f"Input : {x} => Output: {forward_propogation(x[0],x[1])}")
```
